```python
%load_ext autoreload
%autoreload 2


import matplotlib.pyplot as plt
import pandas as pd
```

```python

def read_intra_episode_file(file_name:str) -> pd.DataFrame:
    df = pd.read_csv(file_name, header=None, names=[
        "Episode step", 
        "Accuracy",
        "Sum abs. bias",
        "Reward",
        "Entropy",
    ])
    return df


file_name = "/home/dave/dev/vast_results/results_intra_episode_training.txt"
#file_name = "/home/dave/dev/vast_results/results_intra_episode_evaluate.txt"

df = read_intra_episode_file(file_name)
df["Episode step"] += 1
print(len(df))
```

```python
metrics = ["Accuracy", "Sum abs. bias", "Reward", "Entropy"]

means = df.groupby("Episode step")[metrics].mean()

print(means.head(10))
```

```python

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)
axes = axes.flatten()

for ax, metric in zip(axes, metrics):
    ax.plot(means.index, means[metric], marker="o")
    ax.set_title(metric)
    ax.set_xlabel("Episode step")
    ax.set_ylabel("Mean")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)
axes = axes.flatten()

for ax, metric in zip(axes, metrics):
    df.boxplot(
        column=metric,
        by="Episode step",
        ax=ax
    )
    
    ax.set_title(metric)
    ax.set_xlabel("Episode step")
    ax.set_ylabel(metric)
    ax.grid(True, axis="y", alpha=0.3)

# Remove pandas' automatic overall title
fig.suptitle("")

plt.tight_layout()
plt.show()
```
