```python
FINE_CLASS_COLOURS = {
    "1,2": "tab:blue",
    "12": "tab:blue",
    "1": "tab:blue",
    "2": "tab:blue",
    "3": "tab:orange",
    "4": "tab:red",
    "5": "tab:purple",
}

CHANCE_COLOUR = "black"
PRETRAINED_COLOUR = "brown"
```

```python
def plot_dfs(dfs, titles, legend_index=0):
    assert len(dfs) == len(titles)

    fig, axes = plt.subplots(
        1,
        len(dfs),
        figsize=(6 * len(dfs), 6),
        sharey=True,
    )

    # plt.subplots returns a single Axes if len(dfs) == 1
    if len(dfs) == 1:
        axes = [axes]

    pretrained_value = 0.311

    for i, (ax, df, title) in enumerate(zip(axes, dfs, titles)):
        for fine_class, group in df.groupby("Fine class"):
            group = group.sort_values("Epoch")

            ax.plot(
                group["Epoch"],
                group["Accuracy"],
                label=str(fine_class),
                color=FINE_CLASS_COLOURS.get(str(fine_class)),
                linewidth=2,
                marker="o",
                markersize=4,
            )

        # Reference levels
        ax.axhline(
            pretrained_value,
            color=PRETRAINED_COLOUR,
            linestyle="--",
            linewidth=1.5,
            label="LTM (pre) 3,4,5",
        )

        ax.axhline(
            0.5,
            color=CHANCE_COLOUR,
            linestyle="-.",
            linewidth=1.5,
            label="Random binary",
        )

        ax.set_title(title, fontsize=16, pad=12)
        ax.set_xlabel("Epoch", fontsize=12)
        ax.set_ylim(0.28, 0.92)

        ax.grid(True, alpha=0.25)

        if i == legend_index:
            ax.legend(
                title="Fine class",
                frameon=True,
                framealpha=1.0,
            )

    # Only need ylabel on the leftmost plot because y-axis is shared
    axes[0].set_ylabel("Accuracy", fontsize=12)

    fig.tight_layout()
    plt.show()
```

```python
def get_final_accuracy(df, cutoff:float = 0.75):
    max_epoch = df['Epoch'].max()
    min_epoch = int(float(max_epoch) * cutoff) +1

    df = df[df['Epoch'].between(min_epoch, max_epoch)]
    df = df.groupby("Fine class", as_index=False)["Accuracy"].mean()
    df["Fine class"] = df["Fine class"].astype(str)
    return df

def get_few_shot_dataset(results_df):
    dataset = {}
    few_shot_labels = ["500", "256", "128", "64", "16", "1"]
    for i, key in enumerate(few_shot_labels):
        final_df = get_final_accuracy(results_df[i])
        dataset[key] = final_df
    return dataset
```

```python
def restructure_results(datasets):
    sample_counts = ["500", "256", "128", "64", "16", "1"]
    fine_classes = ["12", "3", "4", "5"]

    restructured = []

    for dataset in datasets:
        result = {}

        for fine_class in fine_classes:
            result[fine_class] = [
                dataset[sample_count]
                    .loc[
                        dataset[sample_count]["Fine class"] == fine_class,
                        "Accuracy",
                    ]
                    .iloc[0]
                for sample_count in sample_counts
            ]

        restructured.append(result)

    return restructured
```

```python
def plot_few_shot_results(plot_data, descriptions):
    sample_counts = ["500", "256", "128", "64", "16", "1"]

    fig, axes = plt.subplots(
        1,
        len(plot_data),
        figsize=(15, 5),
        sharey=True,
    )

    if len(plot_data) == 1:
        axes = [axes]

    for ax, data, description in zip(axes, plot_data, descriptions):
        for fine_class, accuracies in data.items():
            ax.plot(
                sample_counts,
                accuracies,
                marker="o",
                label=str(fine_class),
                color=FINE_CLASS_COLOURS[fine_class],
            )

        ax.set_title(description)
        ax.set_xlabel("Few-shot samples")
        #ax.set_ylim(0, 1)
        ax.grid(True, alpha=0.3)

        # Reference levels
        pretrained_value = 0.311
        ax.axhline(
            pretrained_value,
            color=PRETRAINED_COLOUR,
            linestyle="--",
            linewidth=1.5,
            label="LTM (pre) 3,4,5",
        )

        ax.axhline(
            0.5,
            color=CHANCE_COLOUR,
            linestyle="-.",
            linewidth=1.5,
            label="Random binary",
        )

    axes[0].set_ylabel("Accuracy")
    axes[-1].legend(title="Fine class")

    fig.tight_layout()
    plt.show()
```

```python
datasets = [
    get_few_shot_dataset(results_3_ltm),
    get_few_shot_dataset(results_4_ltm),
    get_few_shot_dataset(results_5_ltm),
]

plot_data = restructure_results(datasets)

plot_few_shot_results(
    plot_data,
    descriptions=[
        "Trained on fine-class 3",
        "Trained on fine-class 4",
        "Trained on fine-class 5",
    ],
)
```
