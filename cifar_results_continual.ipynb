```python
FINE_CLASS_COLOURS = {
    "1,2": "tab:blue",
    "1": "tab:blue",
    "2": "tab:blue",
    "3": "tab:orange",
    "4": "tab:red",
    "5": "tab:purple",
}

REGIME_CHANGE_COLOUR = "lightgrey"
CHANCE_COLOUR = "black"
PRETRAINED_COLOUR = "brown"
```

```python
def plot_dfs(dfs, titles, legend_index=0, vertical_lines:list[float] = []):
    assert len(dfs) == len(titles)

    fig, axes = plt.subplots(
        1,
        len(dfs),
        figsize=(6 * len(dfs), 6),
        sharey=True,
    )

    # plt.subplots returns a single Axes if len(dfs) == 1
    if len(dfs) == 1:
        axes = [axes]

    pretrained_value = 0.311

    for i, (ax, df, title) in enumerate(zip(axes, dfs, titles)):
        for fine_class, group in df.groupby("Fine class"):
            group = group.sort_values("Epoch")

            ax.plot(
                group["Epoch"],
                group["Accuracy"],
                label=str(fine_class),
                color=FINE_CLASS_COLOURS.get(str(fine_class)),
                linewidth=2,
                marker="o",
                markersize=4,
            )

        # Reference levels
        for x in vertical_lines:
            ax.axvline(
                x,
                color=REGIME_CHANGE_COLOUR,
                linestyle="-",
                linewidth=1.5,
                label=None,
            )

        ax.axhline(
            pretrained_value,
            color=PRETRAINED_COLOUR,
            linestyle="--",
            linewidth=1.5,
            label="LTM (pre) 3,4,5",
        )

        ax.axhline(
            0.5,
            color=CHANCE_COLOUR,
            linestyle="-.",
            linewidth=1.5,
            label="Random binary",
        )

        ax.set_title(title, fontsize=16, pad=12)
        ax.set_xlabel("Epoch", fontsize=12)

        ax.grid(True, alpha=0.25)

        if i == legend_index:
            ax.legend(
                title="Fine class",
                frameon=True,
                framealpha=1.0,
            )

    # Only need ylabel on the leftmost plot because y-axis is shared
    axes[0].set_ylabel("Accuracy", fontsize=12)

    fig.tight_layout()
    plt.show()

descriptions_continual = [
    "Trained on fine-classes 3,4,5",
    "Trained on fine-classes 4,5,3",
    "Trained on fine-classes 5,3,4",
]
descriptions_fine_tuning = [
    "Trained on fine-class 3",
    "Trained on fine-class 4",
    "Trained on fine-class 5",
]
```
